```python
class GameState:
    def __init__(self, score):
        self.score = score

    def heuristic(self):
        return self.score


def heuristic_alpha_beta(state,
                         depth,
                         alpha,
                         beta,
                         maximizing):

    if depth == 0:
        return state.heuristic()

    if maximizing:

        value = float('-inf')

        for next_score in [state.score + 1,
                           state.score + 2]:

            child = GameState(next_score)

            value = max(
                value,
                heuristic_alpha_beta(
                    child,
                    depth-1,
                    alpha,
                    beta,
                    False
                )
            )

            alpha = max(alpha, value)

            if alpha >= beta:
                break

        return value

    else:

        value = float('inf')

        for next_score in [state.score - 1,
                           state.score - 2]:

            child = GameState(next_score)

            value = min(
                value,
                heuristic_alpha_beta(
                    child,
                    depth-1,
                    alpha,
                    beta,
                    True
                )
            )

            beta = min(beta, value)

            if alpha >= beta:
                break

        return value


initial = GameState(10)

result = heuristic_alpha_beta(
    initial,
    3,
    float('-inf'),
    float('inf'),
    True
)

print("Heuristic Alpha Beta =", result)
```
