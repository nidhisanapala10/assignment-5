```python
tourist_places = {
    "Goa": {
        "days": 3,
        "cost": 15000,
        "food": ["Seafood", "Fish Curry"]
    },

    "Hyderabad": {
        "days": 2,
        "cost": 10000,
        "food": ["Biryani", "Haleem"]
    },

    "Jaipur": {
        "days": 2,
        "cost": 12000,
        "food": ["Dal Baati", "Ghewar"]
    }
}


def travel_plan(city, budget):

    if city not in tourist_places:
        return "City not found"

    data = tourist_places[city]

    if budget >= data["cost"]:

        return {
            "Destination": city,
            "Duration": data["days"],
            "Estimated Cost": data["cost"],
            "Food": data["food"]
        }

    return "Budget too low"


print(travel_plan("Goa", 20000))
```
