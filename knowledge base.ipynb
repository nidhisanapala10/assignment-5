```python
import networkx as nx
import matplotlib.pyplot as plt

G = nx.Graph()

G.add_edge("Tourist",
           "Goa",
           relation="visits")

G.add_edge("Goa",
           "Seafood",
           relation="famous_food")

G.add_edge("Goa",
           "Beach",
           relation="attraction")

nx.draw(G,
        with_labels=True)

plt.show()
```
